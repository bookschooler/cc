## Step 1 — 환경 확인 (Setup Check)

아래 셀을 실행하면 시스템이 준비됐는지 확인해줍니다.

```python
import subprocess
import os
import sys

print('=' * 50)
print('DESA 환경 점검')
print('=' * 50)

# 1. .env 파일 확인
env_exists = os.path.exists('../.env')
print(f'\n1. .env 파일: {"✅ 존재함" if env_exists else "❌ 없음 — Step 2를 먼저 실행하세요"}')

# 2. API 키 확인
if env_exists:
    with open('../.env') as f:
        content = f.read()
    has_key = 'ANTHROPIC_API_KEY' in content and 'sk-ant-' in content
    print(f'2. API 키 형식: {"✅ 올바름" if has_key else "⚠️ 확인 필요 (sk-ant-로 시작해야 함)"}')
else:
    print('2. API 키: 확인 불가 (.env 없음)')

# 3. anthropic 패키지 확인
try:
    import anthropic
    print(f'3. anthropic 패키지: ✅ 설치됨 (버전: {anthropic.__version__})')
except ImportError:
    print('3. anthropic 패키지: ❌ 설치 필요')
    print('   → 아래 셀을 실행하세요: !pip install anthropic')

# 4. main.py 확인
main_exists = os.path.exists('../main.py')
print(f'4. main.py: {"✅ 존재함" if main_exists else "❌ 없음"}')

print('\n' + '=' * 50)
if env_exists and has_key and main_exists:
    print('🎉 모든 준비 완료! Step 3으로 이동하세요.')
else:
    print('⚠️ 위에서 ❌ 표시된 항목을 먼저 해결하세요.')
```

## Step 2 — API 키 설정 (처음 시작하는 경우만)

1. https://console.anthropic.com 방문

## Step 3 — Planner 단독 실행 (추천 첫 번째 단계)

아래 셀에서 분析 주제를 바꾼 후 실행하세요.

**추천 주제 (쉬운 순서):**
- `'나의 스터디 일정 최적화'`
- `'유튜브 알고리즘이 시청 시간에 미치는 영향'`
- `'삼성전자 주가 패턴 분析'`

```python
import subprocess
import sys

# 여기서 분析 주제를 바꾸세요!
topic = '나의 하루 시간 관리 패턴 분析'

print(f'🚀 Planner 에이전트 시작: "{topic}"')
print('(완료까지 약 20~30초 걸립니다)\n')
print('-' * 50)

result = subprocess.run(
    [sys.executable, 'main.py', topic, '--agent', 'planner', '--no-interrupt'],
    capture_output=True,
    text=True,
    cwd='..'
)

if result.returncode == 0:
    print(result.stdout)
else:
    print('❌ 오류 발생:')
    print(result.stderr)
```

## 결과 확인

프로젝트 완료 후 결과물이 저장되는 위치:

```python
import os
from pathlib import Path

outputs_dir = Path('../outputs')
if outputs_dir.exists():
    files = list(outputs_dir.rglob('*'))
    if files:
        print('📁 outputs/ 폴더에 있는 파일들:')
        for f in sorted(files):
            if f.is_file():
                size = f.stat().st_size
                print(f'  {f.relative_to(outputs_dir.parent)} ({size:,} bytes)')
    else:
        print('outputs/ 폴더가 비어있어요. 프로젝트를 먼저 실행해주세요.')
else:
    print('outputs/ 폴더가 없어요. 프로젝트를 실행하면 자동으로 생성됩니다.')
```
